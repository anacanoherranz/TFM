```python
import pickle
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from itertools import combinations


labels = ["W1", "W2 ", "W3", "W4"]

all_communities = sorted(groups_df["Community"].dropna().unique())

categories = ["0 links", "1 link", "2 balanced", "2 unbalanced", "3 links", "4 links"]

def classify_pair(link_set, ci, cj):
    """Classify the relationship between community ci and cj.
    link_set: set of (src, dst, sign) tuples for the whole network.
    """
    # collect all directed signed links between ci and cj (both directions)
    relevant = [(s, d, v) for s, d, v in link_set if
                (s == ci and d == cj) or (s == cj and d == ci)]
    n = len(relevant)

    if n == 0:
        return "0 links"
    elif n == 1:
        return "1 link"
    elif n == 2:
        (s1, d1, v1), (s2, d2, v2) = relevant
        same_direction = (s1 == s2 and d1 == d2)
        same_sign = (v1 == v2)
        if same_direction:
            # same direction, necessarily different sign → unbalanced
            return "2 unbalanced"
        else:
            # opposite directions
            if same_sign:
                return "2 balanced"
            else:
                return "2 unbalanced"
    elif n == 3:
        return "3 links"
    else:
        return "4 links"

fractions_per_net = []

for data in datos:
    especies = data["especies"]
    mat = np.array(data["matriz_significancia"])

    # Build signed directed link set at community level
    enlaces = set()
    for r in range(len(especies)):
        sp_r = especies[r]
        if sp_r not in species_to_community:
            continue
        ci = species_to_community[sp_r]
        for c in range(len(especies)):
            if r == c:
                continue
            sp_c = especies[c]
            if sp_c not in species_to_community:
                continue
            cj = species_to_community[sp_c]
            val = mat[r][c]
            if val == 0:
                continue
            enlaces.add((ci, cj, int(val)))

    # Count categories over all unordered pairs of communities
    counts = {cat: 0 for cat in categories}
    pairs = list(combinations(all_communities, 2))
    for ci, cj in pairs:
        cat = classify_pair(enlaces, ci, cj)
        counts[cat] += 1

    total_pairs = len(pairs)
    fractions = {cat: counts[cat] / total_pairs for cat in categories}
    fractions_per_net.append(fractions)

    print(f"Counts: {counts}  (total pairs={total_pairs})")

# --- Plot ---
colors = plt.get_cmap("Dark2").colors
colors=sns.color_palette("husl")
x = np.arange(len(categories))

fig, ax = plt.subplots(figsize=(8, 5))

markers = ["s", "o", "^", "D"]  # círculo, cuadrado, triángulo
sizes = [12, 10, 8, 4]

for i, (fracs, label) in enumerate(zip(fractions_per_net, labels)):
    y = [fracs[cat] for cat in categories]
    ax.plot(x, y, markers[i], color=colors[i], markersize=sizes[i], label=label, zorder=4)


ax.set_xticks(x)
ax.set_xticklabels(categories, fontsize=11)
ax.set_ylabel("Fraction of community pairs", fontsize=12)
ax.set_xlabel("Relationship type", fontsize=12)
ax.set_ylim(0, 0.6)
ax.legend(fontsize=10)
ax.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
#plt.savefig("coarse_link_types_basa.png", dpi=150)
plt.show()
print("Saved: coarse_link_types_basa.png")
```

```python
import random
from copy import deepcopy


def randomize_signed_directed_network(
    edges,
    n_swaps=None,
    max_attempts=None,
    allow_selfloops=False
):

    edges = list(deepcopy(edges))

    if n_swaps is None:
        n_swaps = 10 * len(edges)

    if max_attempts is None:
        max_attempts = 100 * len(edges)

    edge_set = set(edges)

    swaps = 0
    attempts = 0

    while swaps < n_swaps and attempts < max_attempts:

        attempts += 1

        # escoger signo
        sign = random.choice([1, -1])

        sign_edges = [e for e in edges if e[2] == sign]

        if len(sign_edges) < 2:
            continue

        e1, e2 = random.sample(sign_edges, 2)

        a, b, s1 = e1
        c, d, s2 = e2

        # evitar degeneraciones
        if len({a, b, c, d}) < 4:
            continue

        new1 = (a, d, sign)
        new2 = (c, b, sign)

        # evitar self-loops
        if not allow_selfloops:
            if a == d or c == b:
                continue

        # evitar duplicados
        if new1 in edge_set or new2 in edge_set:
            continue

        # hacer swap
        edge_set.remove(e1)
        edge_set.remove(e2)

        edge_set.add(new1)
        edge_set.add(new2)

        edges.remove(e1)
        edges.remove(e2)

        edges.append(new1)
        edges.append(new2)

        swaps += 1

    print(f"Realizados {swaps} swaps en {attempts} intentos.")

    return set(edges)
```

```python
import random
from copy import deepcopy


def randomize_signed_directed_network(
    edges,
    n_swaps=None,
    max_attempts=None,
    allow_selfloops=False
):
    """
    Randomiza una red dirigida con signo preservando exactamente:
        - k_out+
        - k_in+
        - k_out-
        - k_in-

    Parámetros
    ----------
    edges : iterable de tuples
        (source, target, sign)

    n_swaps : int
        Número de swaps exitosos deseados

    max_attempts : int
        Máximo número de intentos

    allow_selfloops : bool
        Permitir self-loops o no

    Returns
    -------
    frozenset
        Red randomizada
    """

    # -------------------------
    # Copia local
    # -------------------------

    edge_set = set(deepcopy(edges))

    pos_edges = [e for e in edge_set if e[2] > 0]
    neg_edges = [e for e in edge_set if e[2] < 0]

    L = len(edge_set)

    if n_swaps is None:
        n_swaps = 10 * L

    if max_attempts is None:
        max_attempts = 100 * L

    swaps = 0
    attempts = 0

    # -------------------------
    # MCMC rewiring
    # -------------------------

    while swaps < n_swaps and attempts < max_attempts:

        attempts += 1

        # elegir signo
        if random.random() < 0.5:
            edge_list = pos_edges
            sign = 1
        else:
            edge_list = neg_edges
            sign = -1

        if len(edge_list) < 2:
            continue

        # escoger dos edges distintos
        idx1, idx2 = random.sample(range(len(edge_list)), 2)

        e1 = edge_list[idx1]
        e2 = edge_list[idx2]

        a, b, _ = e1
        c, d, _ = e2

        # evitar degeneraciones
        if len({a, b, c, d}) < 4:
            continue

        # propuesta de swap
        new1 = (a, d, sign)
        new2 = (c, b, sign)

        # evitar self-loops
        if not allow_selfloops:
            if a == d or c == b:
                continue

        # evitar duplicados
        if new1 in edge_set or new2 in edge_set:
            continue

        # -------------------------
        # aplicar swap
        # -------------------------

        edge_set.remove(e1)
        edge_set.remove(e2)

        edge_set.add(new1)
        edge_set.add(new2)

        # actualizar listas locales
        edge_list[idx1] = new1
        edge_list[idx2] = new2

        swaps += 1

    return frozenset(edge_set)
```

```python
from collections import defaultdict


def signed_directed_degree(edges):

    deg = defaultdict(lambda: {
        "out_pos": 0,
        "in_pos": 0,
        "out_neg": 0,
        "in_neg": 0
    })

    for u, v, s in edges:

        if s > 0:
            deg[u]["out_pos"] += 1
            deg[v]["in_pos"] += 1

        else:
            deg[u]["out_neg"] += 1
            deg[v]["in_neg"] += 1

    return dict(deg)

orig_deg = signed_directed_degree(red_original)
rand_deg = signed_directed_degree(red_rand)

assert orig_deg == rand_deg
```

```python
def temporalidad(red1, red2):
    """
    Temporalidad entre dos redes dirigidas con signo.

    T = 1 - (|intersección| / |unión|)

    Parámetros
    ----------
    red1, red2 : set o frozenset
        Conjuntos de enlaces:
        (source, target, sign)

    Returns
    -------
    float
    """

    inter = len(red1 & red2)
    union = len(red1 | red2)

    # evitar división por cero
    if union == 0:
        return 0.0

    return 1 - (inter / union)


def temporalidad_no_signo(red1, red2):
    """
    Temporalidad entre dos redes dirigidas sin signo.

    T = 1 - (|intersección| / |unión|)

    Parámetros
    ----------
    red1, red2 : set o frozenset
        Conjuntos de enlaces:
        (source, target, sign)

    Returns
    -------
    float
    """
    red1 = set((u, v) for u, v, s in red1)
    red2 = set((u, v) for u, v, s in red2)

    inter = len(red1 & red2)
    union = len(red1 | red2)

    # evitar división por cero
    if union == 0:
        return 0.0

    return 1 - (inter / union)


def temporalidad_no_signo_no_dir(red1, red2):
    """
    Temporalidad entre dos redes no dirigidas sin signo.

    T = 1 - (|intersección| / |unión|)

    Parámetros
    ----------
    red1, red2 : set o frozenset
        Conjuntos de enlaces:
        (source, target, sign)

    Returns
    -------
    float
    """
    red1 = set(tuple(sorted((u, v))) for u, v, s in red1)
    red2 = set(tuple(sorted((u, v))) for u, v, s in red2)

    inter = len(red1 & red2)
    union = len(red1 | red2)

    # evitar división por cero
    if union == 0:
        return 0.0

    return 1 - (inter / union)
```

```python
from itertools import combinations
import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------------
# temporalidad media entre todas las parejas
# --------------------------------------------------

def temporalidad_media(redes, temporalidad_func):

    vals = []

    for r1, r2 in combinations(redes, 2):

        vals.append(
            temporalidad_func(r1, r2)
        )

    return np.mean(vals)


# --------------------------------------------------
# estadísticas random
# --------------------------------------------------

def temporalidad_random_stats(redes_random,
                              temporalidad_func):

    n_realizaciones = len(redes_random[0])

    temporales = []

    for k in range(n_realizaciones):

        redes_k = [
            redes_random[i][k]
            for i in range(len(redes_random))
        ]

        t = temporalidad_media(
            redes_k,
            temporalidad_func
        )

        temporales.append(t)

    temporales = np.array(temporales)

    return (
        np.mean(temporales),
        np.std(temporales),
        temporales
    )


# --------------------------------------------------
# calcular todo
# --------------------------------------------------

metodos = {
    "Dir + Signo": temporalidad,
    "Dir sin Signo": temporalidad_no_signo,
    "No Dir sin Signo": temporalidad_no_signo_no_dir
}

labels = []
orig_vals = []
rand_means = []
rand_stds = []

for nombre, func in metodos.items():

    # original
    t_orig = temporalidad_media(
        redes_dir,
        func
    )

    # random
    t_rand_mean, t_rand_std, _ = temporalidad_random_stats(
        redes_random,
        func
    )

    labels.append(nombre)
    orig_vals.append(t_orig)
    rand_means.append(t_rand_mean)
    rand_stds.append(t_rand_std)


# --------------------------------------------------
# gráfica
# --------------------------------------------------

x = np.arange(len(labels))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))

# originales
ax.bar(
    x - width/2,
    orig_vals,
    width,
    color=colors[0],
    label="Original"
)

# random
ax.bar(
    x + width/2,
    rand_means,
    width,
    yerr=rand_stds,
    capsize=5,
    color=colors[1],
    label="Random"
)

ax.set_ylabel("Temporalidad media")
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.set_title("Garba Guracha")

ax.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------------
# pares adyacentes originales
# --------------------------------------------------

def temporalidades_adyacentes(redes,
                              temporalidad_func):

    vals = []

    for i in range(len(redes) - 1):

        t = temporalidad_func(
            redes[i],
            redes[i + 1]
        )

        vals.append(t)

    return np.array(vals)


# --------------------------------------------------
# estadísticas random para pares adyacentes
# --------------------------------------------------

def temporalidades_random_adyacentes(
    redes_random,
    temporalidad_func
):

    n_realizaciones = len(redes_random[0])
    n_pares = len(redes_random) - 1

    # shape:
    # (realización, par)
    vals = np.zeros((n_realizaciones, n_pares))

    for k in range(n_realizaciones):

        for i in range(n_pares):

            r1 = redes_random[i][k]
            r2 = redes_random[i + 1][k]

            vals[k, i] = temporalidad_func(r1, r2)

    mean = vals.mean(axis=0)
    std = vals.std(axis=0)

    return mean, std, vals


# --------------------------------------------------
# plotting
# --------------------------------------------------

metodos = {
    "Dir + Signo": temporalidad,
    "Dir sin Signo": temporalidad_no_signo,
    "No Dir sin Signo": temporalidad_no_signo_no_dir
}

pares_labels = ["1-2", "2-3", "3-4"]

fig, axes = plt.subplots(
    1,
    3,
    figsize=(15, 4),
    sharey=True
)

for ax, (nombre, func) in zip(axes, metodos.items()):

    # originales
    orig = temporalidades_adyacentes(
        redes_dir,
        func
    )

    # random
    rand_mean, rand_std, _ = (
        temporalidades_random_adyacentes(
            redes_random,
            func
        )
    )

    x = np.arange(len(pares_labels))

    # original
    ax.plot(
        x,
        orig,
        marker="o",
        linewidth=2,
        color=colors[0],
        label="Original"
    )

    # random
    ax.errorbar(
        x,
        rand_mean,
        yerr=rand_std,
        marker="o",
        linewidth=2,
        capsize=5,
        color=colors[1],
        label="Random"
    )

    ax.set_title(nombre)

    ax.set_xticks(x)
    ax.set_xticklabels(pares_labels)

    ax.set_xlabel("Par de redes")

axes[0].set_ylabel("Temporalidad")

axes[0].legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------------
# reciprocidad
# --------------------------------------------------

def reciprocidad(red,
                 modo="all"):
    """
    Calcula reciprocidad de una red.

    modos:
        "pos"  -> solo positivos
        "neg"  -> solo negativos
        "all"  -> ignorar signo
    """

    # -------------------------
    # filtrar edges
    # -------------------------

    if modo == "pos":

        edges = set(
            (u, v)
            for u, v, s in red
            if s > 0
        )

    elif modo == "neg":

        edges = set(
            (u, v)
            for u, v, s in red
            if s < 0
        )

    elif modo == "all":

        edges = set(
            (u, v)
            for u, v, s in red
        )

    else:
        raise ValueError("modo inválido")

    # -------------------------
    # reciprocidad
    # -------------------------

    if len(edges) == 0:
        return 0.0

    rec = 0

    for u, v in edges:

        if (v, u) in edges:
            rec += 1

    # cada par recíproco cuenta dos veces
    return rec / len(edges)
```

```python
def reciprocidades_redes(redes,
                         modo):

    return np.array([
        reciprocidad(r, modo)
        for r in redes
    ])

def reciprocidades_random(
    redes_random,
    modo
):

    n_redes = len(redes_random)
    n_random = len(redes_random[0])

    mean = []
    std = []

    for i in range(n_redes):

        vals = np.array([
            reciprocidad(
                redes_random[i][k],
                modo
            )
            for k in range(n_random)
        ])

        mean.append(vals.mean())
        std.append(vals.std())

    return np.array(mean), np.array(std)
```

```python
import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------------
# matriz de categorías desde red
# --------------------------------------------------

def matriz_tipo_pares(red):

    nodos = sorted(
        set(u for u, _, _ in red) |
        set(v for _, v, _ in red)
    )

    idx = {n: i for i, n in enumerate(nodos)}

    n = len(nodos)

    pos = np.zeros((n, n), dtype=int)
    neg = np.zeros((n, n), dtype=int)

    for u, v, s in red:

        i = idx[u]
        j = idx[v]

        if s > 0:
            pos[i, j] = 1

        elif s < 0:
            neg[i, j] = 1

    M = np.zeros((n, n), dtype=int)

    for i in range(n):
        for j in range(i + 1, n):

            pos_ij, neg_ij = pos[i, j], neg[i, j]
            pos_ji, neg_ji = pos[j, i], neg[j, i]

            hay_ij = pos_ij or neg_ij
            hay_ji = pos_ji or neg_ji

            # sin conexión
            if not (hay_ij or hay_ji):

                M[i, j] = M[j, i] = -2

            # no recíproco / mixto
            elif not (hay_ij and hay_ji):

                M[i, j] = M[j, i] = 0

            else:

                # reciprocidad completa
                if pos_ij and neg_ij and pos_ji and neg_ji:

                    M[i, j] = M[j, i] = 2

                # reciprocidad positiva
                elif pos_ij and pos_ji:

                    M[i, j] = M[j, i] = 1

                # reciprocidad negativa
                elif neg_ij and neg_ji:

                    M[i, j] = M[j, i] = -1

                else:

                    M[i, j] = M[j, i] = 0

    return M
```

```python
def categorias_frecuencia(M):

    n = M.shape[0]

    conteo = {
        -2: 0,
        -1: 0,
         0: 0,
         1: 0,
         2: 0
    }

    total = 0

    for i in range(n):
        for j in range(i + 1, n):

            conteo[M[i, j]] += 1
            total += 1

    return {
        k: v / total
        for k, v in conteo.items()
    }
```

```python
import numpy as np
import matplotlib.pyplot as plt

N_ITER = 1000

# --- construir presencia de signos por comunidad ---
def construir_presencia_signo(mat_sig, especies, species_to_community):
    comunidades = sorted(set(
        species_to_community[sp] for sp in especies if sp in species_to_community
    ))
    idx = {c: i for i, c in enumerate(comunidades)}
    n = len(comunidades)
    pos = np.zeros((n, n), dtype=int)
    neg = np.zeros((n, n), dtype=int)
    
    for i, sp_i in enumerate(especies):
        if sp_i not in species_to_community:
            continue
        ci = idx[species_to_community[sp_i]]
        for j, sp_j in enumerate(especies):
            if sp_j not in species_to_community:
                continue
            if i == j:
                continue
            cj = idx[species_to_community[sp_j]]
            val = mat_sig[i][j]
            if val == 1:
                pos[ci, cj] = 1
            elif val == -1:
                neg[ci, cj] = 1
    
    return pos, neg

# --- calcular fracciones sobre pares recíprocos ---
def reciprocidad_por_signo(pos, neg):
    n = pos.shape[0]
    total_rec = 0
    rec_pos = 0  # pares recíprocos donde hay al menos un enlace positivo en ambas direcciones
    rec_neg = 0  # pares recíprocos donde hay al menos un enlace negativo en ambas direcciones
    
    for i in range(n):
        for j in range(i+1, n):
            hay_ij = pos[i,j] or neg[i,j]
            hay_ji = pos[j,i] or neg[j,i]
            
            if hay_ij and hay_ji:
                total_rec += 1
                if pos[i,j] and pos[j,i]:
                    rec_pos += 1  # positivo recíproco
                if neg[i,j] and neg[j,i]:  # 'if' no 'elif' -> no excluyente
                    rec_neg += 1  # negativo recíproco
    
    if total_rec == 0:
        return np.nan, np.nan
    return rec_pos / total_rec, rec_neg / total_rec
# --- generar red aleatoria con misma densidad pos/neg y calcular fracciones ---
def reciprocidad_aleatoria_media(pos, neg, n_iter=N_ITER):
    """
    Aleatoriza manteniendo el número de enlaces positivos y negativos,
    redistribuyéndolos independientemente (pueden solaparse en el mismo par).
    """
    n = pos.shape[0]
    pares_dirigidos = [(i, j) for i in range(n) for j in range(n) if i != j]
    n_pares = len(pares_dirigidos)
    
    n_pos = int(pos.sum())
    n_neg = int(neg.sum())
    
    if n_pos > n_pares or n_neg > n_pares:
        return np.nan, np.nan
    
    vals_pos, vals_neg = [], []
    
    for _ in range(n_iter):
        # Sortear pos y neg independientemente — pueden solaparse
        idx_pos = np.random.choice(n_pares, n_pos, replace=False)
        idx_neg = np.random.choice(n_pares, n_neg, replace=False)
        
        pos_al = np.zeros((n, n), dtype=int)
        neg_al = np.zeros((n, n), dtype=int)
        
        for k in idx_pos:
            i, j = pares_dirigidos[k]
            pos_al[i, j] = 1
        for k in idx_neg:
            i, j = pares_dirigidos[k]
            neg_al[i, j] = 1
        
        fp, fn = reciprocidad_por_signo(pos_al, neg_al)
        if not np.isnan(fp):
            vals_pos.append(fp)
            vals_neg.append(fn)
    
    return (np.mean(vals_pos) if vals_pos else np.nan,
            np.mean(vals_neg) if vals_neg else np.nan)

# --- aplicar a tus redes ---
frac_pos, frac_neg = [], []
frac_aleat_pos, frac_aleat_neg = [], []

for data in datos:
    especies = data["especies"]
    mat_sig  = data["matriz_significancia"]
    
    pos, neg = construir_presencia_signo(mat_sig, especies, species_to_community)
    
    fp, fn = reciprocidad_por_signo(pos, neg)
    frac_pos.append(fp)
    frac_neg.append(fn)
    
    fp_al, fn_al = reciprocidad_aleatoria_media(pos, neg)
    frac_aleat_pos.append(fp_al)
    frac_aleat_neg.append(fn_al)

# Normalizar
frac_pos      = np.array(frac_pos)
frac_neg      = np.array(frac_neg)
frac_aleat_pos = np.array(frac_aleat_pos)
frac_aleat_neg = np.array(frac_aleat_neg)

frac_pos_norm = frac_pos / frac_aleat_pos
frac_neg_norm = frac_neg / frac_aleat_neg

# --- plot ---
x = np.arange(1, len(frac_pos) + 1)
plt.figure(figsize=(8, 5))
plt.plot(x, frac_pos_norm, marker='o', linewidth=2, label='Reciprocidad positiva (norm.)')
plt.plot(x, frac_neg_norm, marker='o', linewidth=2, label='Reciprocidad negativa (norm.)')
plt.axhline(1, color='gray', linestyle='--', linewidth=1, label='Esperado aleatorio')
plt.xticks(x, ["w1", "w2", "w3", "w4"])
plt.xlabel("Ventana temporal")
plt.ylabel("Fracción observada / esperada aleatoria")
plt.title("Tipo de reciprocidad por signo (normalizada), Garba Guracha con pc")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from scipy.cluster.hierarchy import linkage, leaves_list
from scipy.spatial.distance import pdist
from matplotlib.lines import Line2D

# =========================
# PRESENCIA DE SIGNO
# =========================
def construir_presencia_signo(mat_sig, especies, species_to_community):
    comunidades = sorted(set(
        species_to_community[sp] for sp in especies if sp in species_to_community
    ))
    
    idx = {c: i for i, c in enumerate(comunidades)}
    n = len(comunidades)
    
    pos = np.zeros((n, n), dtype=int)
    neg = np.zeros((n, n), dtype=int)
    
    for i, sp_i in enumerate(especies):
        if sp_i not in species_to_community:
            continue
        ci = idx[species_to_community[sp_i]]
        
        for j, sp_j in enumerate(especies):
            if sp_j not in species_to_community:
                continue
            if i == j:
                continue
            
            cj = idx[species_to_community[sp_j]]
            val = mat_sig[i][j]
            
            if val == 1:
                pos[ci, cj] = 1
            elif val == -1:
                neg[ci, cj] = 1
    
    return pos, neg, comunidades


# =========================
# MATRIZ TIPO PARES
# =========================
def matriz_tipo_pares_v3(mat_sig, especies, species_to_community):
    pos, neg, comunidades = construir_presencia_signo(
        mat_sig, especies, species_to_community
    )
    
    n = len(comunidades)
    M = np.zeros((n, n))
    
    for i in range(n):
        for j in range(i+1, n):
            
            pos_ij, neg_ij = pos[i,j], neg[i,j]
            pos_ji, neg_ji = pos[j,i], neg[j,i]
            
            hay_ij = pos_ij or neg_ij
            hay_ji = pos_ji or neg_ji
            
            if not (hay_ij or hay_ji):
                M[i,j] = M[j,i] = -2
            
            elif not (hay_ij and hay_ji):
                M[i,j] = M[j,i] = 0
            
            else:
                if pos_ij and neg_ij and pos_ji and neg_ji:
                    M[i,j] = M[j,i] = 2
                elif pos_ij and pos_ji:
                    M[i,j] = M[j,i] = 1
                elif neg_ij and neg_ji:
                    M[i,j] = M[j,i] = -1
                else:
                    M[i,j] = M[j,i] = 0
    
    return M, comunidades


# =========================
# ORDEN FIJO (usando w1)
# =========================
M_ref, comunidades = matriz_tipo_pares_v3(
    datos[0]["matriz_significancia"],
    datos[0]["especies"],
    species_to_community
)

dist = pdist(M_ref, metric='euclidean')
Z = linkage(dist, method='average')
order = leaves_list(Z)

comunidades_ord = [comunidades[i] for i in order]

# =========================
# GENERAR MATRICES ORDENADAS
# =========================
matrices = []

for data in datos:
    M, comunidades = matriz_tipo_pares_v3(
        data["matriz_significancia"],
        data["especies"],
        species_to_community
    )
    
    M_ord = M[order][:, order]
    matrices.append(M_ord)

titulos = ["w1", "w2", "w3", "w4"]

# =========================
# COLORES
# =========================
base = plt.cm.RdYlGn

gray = (0.85, 0.85, 0.85)
red = base(0.0)
yellow = base(0.5)
green = base(1.0)

colores = [gray, red, yellow, green, yellow]

cmap = mcolors.ListedColormap(colores)
bounds = [-2.5, -1.5, -0.5, 0.5, 1.5, 2.5]
norm = mcolors.BoundaryNorm(bounds, cmap.N)

# =========================
# PLOT
# =========================
fig, axes = plt.subplots(1, 4, figsize=(20, 5))

for k, ax in enumerate(axes):
    
    M = matrices[k]
    
    # máscara: ocultar triángulo SUPERIOR
    tri_mask = np.triu(np.ones_like(M, dtype=bool))
    M_plot = np.ma.masked_where(tri_mask, M)
    
    im = ax.imshow(M_plot, cmap=cmap, norm=norm)
    
    # etiquetas
    ax.set_xticks(range(len(comunidades_ord)))
    ax.set_xticklabels(comunidades_ord, rotation=90, fontsize=12)
    
    ax.set_yticks(range(len(comunidades_ord)))
    ax.set_yticklabels(comunidades_ord, fontsize=12)
    
    ax.set_title(titulos[k])
    
    # =========================
    # GRID SOLO EN TRIÁNGULO INFERIOR
    # =========================
    n = len(comunidades_ord)
    for i in range(n):
        for j in range(n):
            if i > j:  # solo inferior
                ax.add_patch(plt.Rectangle((j-0.5, i-0.5), 1, 1,
                               fill=False, edgecolor='gray', linewidth=0.5))
    
    # =========================
    # HATCH PARA ++--
    # =========================
    mask_2 = (M == 2) & (~tri_mask)
    y, x = np.where(mask_2)
    
    for xi, yi in zip(x, y):
        # líneas rojas
        for offset in np.linspace(-0.4, 0.4, 6):
            ax.plot([xi + offset, xi + offset],
                    [yi - 0.45, yi + 0.45],
                    color=red, linewidth=1.5)
        
        # líneas verdes
        for offset in np.linspace(-0.4, 0.4, 6):
            ax.plot([xi - 0.45, xi + 0.45],
                    [yi + offset, yi + offset],
                    color=green, linewidth=1.5)

# ocultar eje Y en 2 y 3
axes[1].set_yticks([])
axes[2].set_yticks([])
axes[3].set_yticks([])

# =========================
# LEYENDA
# =========================
legend_elements = [
    Line2D([0], [0], marker='s', color='w', label='Sin conexión',
           markerfacecolor=gray, markersize=15),
    Line2D([0], [0], marker='s', color='w', label='No recíproco / mixto (+/-)',
           markerfacecolor=yellow, markersize=15),
    Line2D([0], [0], marker='s', color='w', label='Reciprocidad negativa',
           markerfacecolor=red, markersize=15),
    Line2D([0], [0], marker='s', color='w', label='Reciprocidad positiva',
           markerfacecolor=green, markersize=15),
    Line2D([0], [0], marker='s', color='w', label='Reciprocidad completa (++--)',
           markerfacecolor=yellow, markersize=15)
]

fig.legend(handles=legend_elements, loc='lower center',
           bbox_to_anchor=(0.5, -0.17), ncol=5, fontsize=12)

fig.suptitle("Garba Guracha con pc")

plt.tight_layout()
plt.show()
```

```python
N_ITER = 1000

def categorias_frecuencia(M):
    """Cuenta frecuencia de cada categoría en el triángulo inferior."""
    n = M.shape[0]
    conteo = {-2: 0, -1: 0, 0: 0, 1: 0, 2: 0}
    total = 0
    for i in range(n):
        for j in range(i+1, n):
            conteo[M[i,j]] += 1
            total += 1
    return {k: v/total for k, v in conteo.items()} if total > 0 else conteo

def frecuencias_aleatorias_media(pos, neg, n_iter=N_ITER):
    """Media de frecuencias de categorías sobre n_iter redes aleatorias."""
    n = pos.shape[0]
    pares_dirigidos = [(i, j) for i in range(n) for j in range(n) if i != j]
    n_pares = len(pares_dirigidos)
    n_pos = int(pos.sum())
    n_neg = int(neg.sum())

    acum = {-2: [], -1: [], 0: [], 1: [], 2: []}

    for _ in range(n_iter):
        idx_pos = np.random.choice(n_pares, n_pos, replace=False)
        idx_neg = np.random.choice(n_pares, n_neg, replace=False)

        pos_al = np.zeros((n, n), dtype=int)
        neg_al = np.zeros((n, n), dtype=int)
        for k in idx_pos:
            i, j = pares_dirigidos[k]
            pos_al[i, j] = 1
        for k in idx_neg:
            i, j = pares_dirigidos[k]
            neg_al[i, j] = 1

        # Reutilizar matriz_tipo_pares_v3 pero pasando pos/neg directamente
        M_al = _matriz_desde_pos_neg(pos_al, neg_al)
        freqs = categorias_frecuencia(M_al)
        for cat, v in freqs.items():
            acum[cat].append(v)

    return {cat: np.mean(vals) for cat, vals in acum.items()}

def _matriz_desde_pos_neg(pos, neg):
    """Construye la matriz de categorías directamente desde pos y neg."""
    n = pos.shape[0]
    M = np.zeros((n, n))
    for i in range(n):
        for j in range(i+1, n):
            pos_ij, neg_ij = pos[i,j], neg[i,j]
            pos_ji, neg_ji = pos[j,i], neg[j,i]
            hay_ij = pos_ij or neg_ij
            hay_ji = pos_ji or neg_ji
            if not (hay_ij or hay_ji):
                M[i,j] = M[j,i] = -2
            elif not (hay_ij and hay_ji):
                M[i,j] = M[j,i] = 0
            else:
                if pos_ij and neg_ij and pos_ji and neg_ji:
                    M[i,j] = M[j,i] = 2
                elif pos_ij and pos_ji:
                    M[i,j] = M[j,i] = 1
                elif neg_ij and neg_ji:
                    M[i,j] = M[j,i] = -1
                else:
                    M[i,j] = M[j,i] = 0
    return M

# --- Calcular frecuencias observadas y aleatorias por ventana ---
categorias = [-2, -1, 0, 1, 2]
labels_cat = {-2: 'Sin conexión', -1: 'Rec. negativa', 0: 'No rec./mixto',
               1: 'Rec. positiva', 2: 'Rec. completa'}
colores_cat = {-2: gray, -1: red, 0: "black", 1: green, 2: 'purple'}

freqs_obs   = []  # frecuencias observadas por ventana
freqs_aleat = []  # frecuencias aleatorias por ventana

for data in datos:
    pos, neg, _ = construir_presencia_signo(
        data["matriz_significancia"], data["especies"], species_to_community)
    M = _matriz_desde_pos_neg(pos, neg)
    freqs_obs.append(categorias_frecuencia(M))
    freqs_aleat.append(frecuencias_aleatorias_media(pos, neg))

# --- Plot: ratio observado/aleatorio por categoría y ventana ---
x = np.arange(1, len(datos) + 1)
fig, ax = plt.subplots(figsize=(9, 5))

for cat in categorias:
    obs   = np.array([f[cat] for f in freqs_obs])
    aleat = np.array([f[cat] for f in freqs_aleat])
    ratio = obs / np.where(aleat == 0, np.nan, aleat)
    ax.plot(x, ratio, marker='o', linewidth=2,
            label=labels_cat[cat], color=colores_cat[cat])

ax.axhline(1, color='gray', linestyle='--', linewidth=1, label='Esperado aleatorio')
ax.set_xticks(x)
ax.set_xticklabels(["w1", "w2", "w3", "w4"])
ax.set_xlabel("Ventana temporal")
ax.set_ylabel("Frecuencia observada / esperada aleatoria")
ax.set_title("Tipo de par (normalizado por aleatorio), Garba Guracha con pc")
ax.legend(bbox_to_anchor=(1.01, 1), loc='upper left')
ax.grid(True)
plt.tight_layout()
plt.show()
```

```python
# construir A a nivel de especies (tu función original)
def construir_A_especies(mat_sig):
    n = len(mat_sig)
    A = np.zeros((n, n))
    
    for i in range(n):
        for j in range(i+1, n):
            a_ij = mat_sig[i][j]
            a_ji = mat_sig[j][i]

            if a_ij == 0 and a_ji == 0:
                continue

            if a_ij * a_ji == -1:
                continue

            sign = a_ij if a_ij != 0 else a_ji
            A[i, j] = sign
            A[j, i] = sign

    return A

resultados_comunidades = {}

for data in datos:
    especies = data["especies"]
    mat_sig = data["matriz_significancia"]
    
    A = construir_A_especies(mat_sig)
    
    k_local = calcular_k_local(A)
    SC = calcular_SC(A)
    
    # agrupar por comunidad
    for i, sp in enumerate(especies):
        if sp not in species_to_community:
            continue
        
        com = species_to_community[sp]
        
        if com not in resultados_comunidades:
            resultados_comunidades[com] = {
                "k": [],
                "SC": []
            }
        
        resultados_comunidades[com]["k"].append(k_local[i])
        resultados_comunidades[com]["SC"].append(SC[i])

k_com = {}
SC_com = {}

for com, vals in resultados_comunidades.items():
    k_com[com] = np.nanmean(vals["k"])
    SC_com[com] = np.nanmean(vals["SC"])

```

```python
balance_medio = []
balance_global = []


for t in range(k_all.shape[0]):
    SC_row = SC_all[t, :]
    k_row = k_all[t, :]
    
    # --- máscara válida ---
    mask = ~np.isnan(SC_row) & ~np.isnan(k_row)
    
    if np.sum(mask) == 0:
        balance_medio.append(np.nan)
        balance_global.append(np.nan)
        continue
    
    SC_valid = SC_row[mask]
    k_valid = k_row[mask]
    
    # --- balance medio ---
    bm = np.mean(k_valid)
    
    # --- balance global (ponderado) ---
    if np.sum(k_valid) == 0:
        bg = np.nan
    else:
        bg = np.sum(SC_valid * k_valid) / np.sum(SC_valid)
    
    balance_medio.append(bm)
    balance_global.append(bg)
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from collections import defaultdict
from networkx.drawing.nx_pydot import pydot_layout
import matplotlib.colors as mcolors

from lib.halfarrowpatch import *

class HalfArrowPatch(FancyArrowPatch):
    """
    A custom arrow patch that draws arrows with half arrowheads.
    Useful for drawing multiple parallel edges between nodes.
    """
    
    def __init__(self, posA, posB, marker='-|>', width=1, height=1, 
                 color='black', linewidth=1, curvature=0.3, 
                 arrow_linewidth=2, **kwargs):
        """
        Initialize a HalfArrowPatch.
        
        Parameters:
        -----------
        posA, posB : tuples
            Starting and ending positions (x, y)
        marker : str
            Arrow style ('-|>' for right half, '|\\' for left half, etc.)
        width, height : float
            Width and height of the arrowhead
        color : str or tuple
            Arrow color
        linewidth : float
            Width of the arrow shaft
        curvature : float
            Curvature of the arrow (0 = straight, >0 = curved)
        arrow_linewidth : float
            Width of the arrowhead lines
        **kwargs : additional arguments for FancyArrowPatch
        """
        # Store parameters
        self.marker = marker
        self.arrow_width = width
        self.arrow_height = height
        self.arrow_linewidth = arrow_linewidth
        self.curvature = curvature
        
        # Initialize the parent class
        super().__init__(posA, posB, 
                        connectionstyle=f"arc3,rad={curvature}",
                        arrowstyle='-',  # Start with no arrowhead, we'll draw it manually
                        color=color,
                        linewidth=linewidth,
                        **kwargs)
        
        self._color = color
        self._posA = np.array(posA)
        self._posB = np.array(posB)
        
    def draw(self, renderer):
        """Override draw method to add custom half arrowhead."""
        # Draw the curved line first
        super().draw(renderer)
        
        # Get the transformed points
        trans = self.get_transform()
        posA_trans = trans.transform(self._posA)
        posB_trans = trans.transform(self._posB)
        
        # Calculate the direction at the end of the curve
        # For curved arrows, we need the tangent at the endpoint
        if self.curvature != 0:
            # For curved arrows, calculate tangent at endpoint
            C, R = self._get_circle_center()
            if C is not None:
                # Vector from center to endpoint
                v = posB_trans - C
                # Tangent is perpendicular to radius
                tangent = np.array([-v[1], v[0]])
                tangent = tangent / np.linalg.norm(tangent)
            else:
                # Fallback to straight line direction
                tangent = posB_trans - posA_trans
                tangent = tangent / np.linalg.norm(tangent)
        else:
            # Straight line - direction is from start to end
            tangent = posB_trans - posA_trans
            tangent = tangent / np.linalg.norm(tangent)
        
        # Perpendicular vector (90 degrees rotation)
        perp = np.array([-tangent[1], tangent[0]])
        
        # Calculate arrowhead points based on marker style
        arrowhead_center = posB_trans
        
        if self.marker == '-|>' or self.marker == '|>':
            # Right half arrowhead (pointing to the right of the direction)
            points = [
                arrowhead_center,
                arrowhead_center - tangent * self.arrow_height + perp * self.arrow_width/2,
                arrowhead_center - tangent * self.arrow_height
            ]
        elif self.marker == '|\\' or self.marker == '<|':
            # Left half arrowhead (pointing to the left of the direction)
            points = [
                arrowhead_center,
                arrowhead_center - tangent * self.arrow_height - perp * self.arrow_width/2,
                arrowhead_center - tangent * self.arrow_height
            ]
        elif self.marker == '|<':
            # Half arrowhead on the other side (for bidirectional)
            points = [
                arrowhead_center,
                arrowhead_center - tangent * self.arrow_height + perp * self.arrow_width/2,
                arrowhead_center - tangent * self.arrow_height
            ]
        else:
            # Default full arrowhead
            points = [
                arrowhead_center,
                arrowhead_center - tangent * self.arrow_height + perp * self.arrow_width/2,
                arrowhead_center - tangent * self.arrow_height - perp * self.arrow_width/2,
                arrowhead_center
            ]
        
        # Create the arrowhead patch
        if len(points) >= 3:
            # Create path for the filled arrowhead
            verts = points
            codes = [Path.MOVETO] + [Path.LINETO] * (len(points)-2) + [Path.CLOSEPOLY]
            
            path = Path(verts, codes)
            patch = PathPatch(path, facecolor=self._color, 
                            edgecolor=self._color, 
                            linewidth=self.arrow_linewidth,
                            transform=transforms.IdentityTransform(),
                            zorder=self.zorder+1)
            
            # Draw the arrowhead
            patch.draw(renderer)
    
    def _get_circle_center(self):
        """Calculate the center of the circle for curved arrows."""
        try:
            A = self._posA
            B = self._posB
            rad = self.curvature
            
            # Calculate the center of the arc
            M = (A + B) / 2
            d = B - A
            L = np.linalg.norm(d)
            
            if L == 0:
                return None, None
                
            n = np.array([-d[1], d[0]]) / L
            h = rad * L / 2
            R = (L**2 / 4 + h**2) / (2 * h)
            C = M + n * (R - L**2 / (8 * h))
            
            return C, R
        except:
            return None, None


def test_half_arrow():
    """Test function to demonstrate HalfArrowPatch usage."""
    fig, ax = plt.subplots(figsize=(10, 10))
    
    # Create some test positions
    pos = {
        'A': np.array([0, 0]),
        'B': np.array([1, 0.5]),
        'C': np.array([0.5, 1]),
        'D': np.array([-0.5, 0.5])
    }
    
    # Draw nodes
    for node, p in pos.items():
        circle = plt.Circle(p, 0.1, color='lightblue', ec='black', lw=2)
        ax.add_patch(circle)
        ax.text(p[0], p[1]+0.15, node, ha='center', fontsize=12)
    
    # Test different arrow styles
    # Right half arrow
    arrow1 = HalfArrowPatch(
        pos['A'], pos['B'],
        marker='-|>',
        width=0.2, height=0.3,
        color='blue',
        linewidth=2,
        curvature=0.2,
        arrow_linewidth=2
    )
    ax.add_patch(arrow1)
    
    # Left half arrow
    arrow2 = HalfArrowPatch(
        pos['A'], pos['C'],
        marker='|\\',
        width=0.2, height=0.3,
        color='red',
        linewidth=2,
        curvature=0.2,
        arrow_linewidth=2
    )
    ax.add_patch(arrow2)
    
    # Double arrows (parallel)
    arrow3 = HalfArrowPatch(
        pos['A'], pos['D'],
        marker='-|>',
        width=0.2, height=0.3,
        color='green',
        linewidth=2,
        curvature=0.15,
        arrow_linewidth=2
    )
    ax.add_patch(arrow3)
    
    arrow4 = HalfArrowPatch(
        pos['A'], pos['D'],
        marker='|\\',
        width=0.2, height=0.3,
        color='orange',
        linewidth=2,
        curvature=0.15,
        arrow_linewidth=2
    )
    # Adjust offset manually
    arrow4._posA = arrow4._posA + np.array([0.05, 0.05])
    arrow4._posB = arrow4._posB + np.array([0.05, 0.05])
    ax.add_patch(arrow4)
    
    ax.set_xlim(-1, 2)
    ax.set_ylim(-1, 2)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.set_title('HalfArrowPatch Test')
    
    plt.show()


if __name__ == "__main__":
    test_half_arrow()
    

```

```python
import numpy as np
import pandas as pd
import pickle
import networkx as nx
import matplotlib.pyplot as plt
from collections import defaultdict
import matplotlib.patheffects as path_effects
import matplotlib.colors as mcolors

from lib.halfarrowpatch import *

# =====================================================
# FUNCIONES AUXILIARES
# =====================================================

def arc3_circle_center(posA, posB, rad):
    A, B = np.array(posA), np.array(posB)
    M = (A + B) / 2
    d = B - A
    L = np.linalg.norm(d)
    n = np.array([-d[1], d[0]]) / L
    h = rad * L / 2
    R = (L**2 / 4 + h**2) / (2 * h)
    C = M + n * (R - L**2 / (8 * h))
    return C, R


def rotate_point(p, center, angle):
    s, c = np.sin(angle), np.cos(angle)
    v = p - center
    return center + np.array([
        c*v[0] - s*v[1],
        s*v[0] + c*v[1]
    ])


def add_curved_edge(
    ax,
    pos,
    u,
    v,
    color="black",
    weight=1.5,
    rad=0.3,
    angle=0,
    offset=0.0,
    node_radius_u=0.05,
    node_radius_v=0.05,
    arrowstyle='-|>',
    zorder=1,
    **kwargs
):

    U = np.array(pos[u])
    V = np.array(pos[v])

    C, R = arc3_circle_center(U, V, rad)

    U2 = rotate_point(U, C, np.deg2rad(angle))
    V2 = rotate_point(V, C, np.deg2rad(angle))

    x1, y1 = U2
    x2, y2 = V2

    dx, dy = x2 - x1, y2 - y1
    dist = np.sqrt(dx**2 + dy**2)

    dir_vec = np.array([dx, dy]) / dist
    perp = np.array([-dy, dx]) / dist

    posA = (
        np.array([x1, y1])
        + perp * offset
        + dir_vec * np.sqrt(node_radius_u)
    )

    posB = (
        np.array([x2, y2])
        + perp * offset
        - dir_vec * np.sqrt(node_radius_v)
    )

    arrow = HalfArrowPatch(
        posA=posA,
        posB=posB,
        marker=arrowstyle,
        width=1,
        height=1,
        color=color,
        linewidth=weight,
        curvature=rad,
        zorder=zorder,
        arrow_linewidth=2.5*np.sqrt(weight),
        **kwargs
    )

    ax.add_patch(arrow)


def sequential_custom_generator():

    colors_list = [
        "#f6ad02",
        "#dbb314",
        "#a1a94a",
        "#88a559",
        "#6ea067",
        "#449d4c",
        "#199a31",
        "#0d841d"
    ]

    np.random.seed(42)
    random_indices = np.random.permutation(len(colors_list))

    idx = 0

    while True:
        yield mcolors.to_rgb(
            colors_list[random_indices[idx] % len(colors_list)]
        )
        idx += 1


# =====================================================
# CONFIG
# =====================================================

windows_names = ["w1", "w2", "w3", "w4"]

POS_COLOR = "#0F4C5C"
NEG_COLOR = "#E36414"

community_to_index = {
    'Afroalpine': 4,
    'Ericaceous belt': 1,
    'Upper forest limit': 7,
    'Forest belt': 3,
    'Lower forest limit': 0,
    'Ruderal': 5,
    'Grassland': 6,
    'Poaceae': 2
}

labels = {
    4: 'Afroalpine',
    1: 'Ericaceous\nbelt',
    7: 'Upper forest\nlimit',
    3: 'Forest belt',
    0: 'Lower forest\nlimit',
    5: 'Ruderal',
    6: 'Grassland',
    2: 'Poaceae'
}

order = {
    'Afroalpine': 5,
    'Ericaceous\nbelt': 2,
    'Upper forest\nlimit': 8,
    'Forest belt': 4,
    'Lower forest\nlimit': 0,
    'Ruderal': 6,
    'Grassland': 7,
    'Poaceae': 3
}

# =====================================================
# FIGURA
# =====================================================

fig, axes = plt.subplots(1, 4, figsize=(60, 20))

# =====================================================
# LOOP VENTANAS
# =====================================================

for idx, window_name in enumerate(windows_names):

    print(f"Procesando {window_name}")

    ax = axes[idx]

    # =================================================
    # CARGAR DATOS
    # =================================================

    with open(f"resultado_{window_name}.pkl", "rb") as f:
        data = pickle.load(f)

    sig = np.array(data["matriz_significancia"])
    communities = data["especies"]

    n_communities = len(communities)

    # =================================================
    # RED
    # =================================================

    G = nx.MultiDiGraph()

    for comm in communities:
        G.add_node(comm, size=1)

    # =================================================
    # EDGES
    # =================================================

    for i in range(n_communities):

        for j in range(n_communities):

            if i == j:
                continue

            sign = sig[i, j]

            if sign == 0:
                continue

            u = communities[i]
            v = communities[j]

            coef = data["matriz_coeficientes"][i, j]

            # evitar weights extremadamente pequeños
            weight = np.abs(coef)

            # opcional: escala visual
            weight = 1 + 16 * np.sqrt(weight)

            if sign > 0:

                G.add_edge(
                    u,
                    v,
                    color=POS_COLOR,
                    weight=weight
                )

            elif sign < 0:

                G.add_edge(
                    u,
                    v,
                    color=NEG_COLOR,
                    weight=weight
                )

    # =================================================
    # POSICIONES
    # =================================================

    communities_present = list(G.nodes())

    initial_pos = {}

    center_comm = None

    for comm in communities_present:

        if community_to_index[comm] == 0:

            center_comm = comm
            initial_pos[comm] = np.array([0.0, 0.0])

            break

    if center_comm is not None:

        N = len(communities_present)

        angle_step = 2 * np.pi / (N - 1)

        other_comms = [
            c for c in communities_present
            if c != center_comm
        ]

        for comm in other_comms:

            idx_comm = community_to_index[comm]

            label = labels[idx_comm]

            if label in order:

                angle = angle_step * (order[label] - 2)

            else:

                angle = np.random.uniform(0, 2*np.pi)

            initial_pos[comm] = np.array([
                np.cos(angle),
                np.sin(angle)
            ])

    # =================================================
    # LAYOUT
    # =================================================

    if len(initial_pos) > 0:

        pos = nx.spring_layout(
            G,
            weight='weight',
            seed=1,
            pos=initial_pos,
            iterations=2
        )

    else:

        pos = nx.spring_layout(
            G,
            seed=1
        )

    # =================================================
    # NODOS
    # =================================================

    node_sizes = {
        k: G.nodes[k]['size'] * 4000
        for k in G.nodes()
    }

    community_colors_gen = sequential_custom_generator()

    community_colors = {
        k: next(community_colors_gen)
        for k in G.nodes()
    }

    node_size_scale = 1 / 700000

    scale = 0.00094

    # =================================================
    # AGRUPAR EDGES
    # =================================================

    edge_groups = defaultdict(list)

    for u, v, edge_data in G.edges(data=True):

        edge_groups[(u, v)].append(edge_data)

    max_total_weight = max(
        sum(e['weight'] for e in edges)
        for edges in edge_groups.values()
    ) if edge_groups else 1

    # =================================================
    # DIBUJAR EDGES
    # =================================================

    for (u, v), edges in edge_groups.items():

        n_edges_pair = len(edges)

        total_width = sum(
            edge["weight"]
            for edge in edges
        )

        alpha = min(
            1.1 - np.exp(
                -4 * total_width / max_total_weight
            ),
            1
        )

        # =============================================
        # UNA EDGE
        # =============================================

        if n_edges_pair == 1:

            edge_data = edges[0]

            arrow_style = (
                '|/'
                if edge_data["color"] == POS_COLOR
                else '|\\'
            )

            add_curved_edge(
                ax,
                pos,
                u,
                v,
                color=edge_data["color"],
                weight=edge_data["weight"],
                rad=0.25,
                angle=0,
                offset=0.0,
                node_radius_u=node_sizes[u] * node_size_scale,
                node_radius_v=node_sizes[v] * node_size_scale,
                arrowstyle=arrow_style,
                zorder=-1,
                alpha=alpha
            )

        # =============================================
        # DOS EDGES
        # =============================================

        elif n_edges_pair == 2:

            pos_edges = [
                e for e in edges
                if e["color"] == POS_COLOR
            ]

            neg_edges = [
                e for e in edges
                if e["color"] == NEG_COLOR
            ]

            sorted_edges = []

            if pos_edges:
                sorted_edges.append(pos_edges[0])

            if neg_edges:
                sorted_edges.append(neg_edges[0])

            if len(sorted_edges) == 2:

                w0 = sorted_edges[0]["weight"]
                w1 = sorted_edges[1]["weight"]

                offset = (w0 + w1) * scale

                np.random.seed(0)

                rad = np.random.uniform(0.15, 0.25)

                # positiva

                add_curved_edge(
                    ax,
                    pos,
                    u,
                    v,
                    color=sorted_edges[0]["color"],
                    weight=w0,
                    rad=rad,
                    angle=0,
                    offset=-offset - 0.02,
                    node_radius_u=node_sizes[u] * node_size_scale,
                    node_radius_v=node_sizes[v] * node_size_scale,
                    arrowstyle='|/',
                    zorder=-1,
                    alpha=alpha
                )

                # negativa

                add_curved_edge(
                    ax,
                    pos,
                    u,
                    v,
                    color=sorted_edges[1]["color"],
                    weight=w1,
                    rad=rad,
                    angle=0,
                    offset=-0.02,
                    node_radius_u=node_sizes[u] * node_size_scale,
                    node_radius_v=node_sizes[v] * node_size_scale,
                    arrowstyle='|\\',
                    zorder=-1,
                    alpha=alpha
                )

    # =================================================
    # DIBUJAR NODOS
    # =================================================

    nx.draw_networkx_nodes(
        G,
        pos,
        node_color=[
            community_colors[node]
            for node in G.nodes()
        ],
        ax=ax,
        node_size=[
            node_sizes[k]
            for k in G.nodes()
        ],
        edgecolors="black",
        linewidths=4
    )

    # =================================================
    # LABELS
    # =================================================

    label_pos = {}

    for k in G.nodes():

        if pos[k][1] > 0.1:

            offset = np.array([
                0,
                np.sqrt(node_sizes[k]) * 0.001 + 0.06
            ])

        elif pos[k][1] < -0.1:

            offset = np.array([
                0,
                -np.sqrt(node_sizes[k]) * 0.001 - 0.06
            ])

        elif pos[k][0] > 0.2:

            offset = np.array([
                0,
                np.sqrt(node_sizes[k]) * 0.001 + 0.06
            ])

        else:

            offset = np.array([0, 0.0])

        label_pos[k] = pos[k] + offset

    node_labels = {}

    for k in G.nodes():

        idx_comm = community_to_index[k]

        node_labels[k] = labels[idx_comm]

    texts = nx.draw_networkx_labels(
        G,
        label_pos,
        ax=ax,
        labels=node_labels,
        font_size=27,
        font_color='black'
    )

    for text in texts.values():

        text.set_path_effects([
            path_effects.Stroke(
                linewidth=1.5,
                foreground='white'
            ),
            path_effects.Normal()
        ])

    # =================================================
    # LIMITES
    # =================================================

    xlim = ax.get_xlim()
    ylim = ax.get_ylim()

    width = xlim[1] - xlim[0]
    height = ylim[1] - ylim[0]

    if width > height:

        center_y = (ylim[0] + ylim[1]) / 2

        ax.set_ylim(
            center_y - width/2,
            center_y + width/2
        )

    else:

        center_x = (xlim[0] + xlim[1]) / 2

        ax.set_xlim(
            center_x - height/2,
            center_x + height/2
        )

    margin = 0.13

    ax.set_ylim(
        ax.get_ylim()[0] - margin,
        ax.get_ylim()[1] + margin
    )

    ax.set_xlim(
        ax.get_xlim()[0] - margin,
        ax.get_xlim()[1] + margin
    )

    # =================================================
    # TEXTO
    # =================================================

    negative_links = len([
        e for e in G.edges(data=True)
        if e[2]['color'] == NEG_COLOR
    ])

    total_links = G.number_of_edges()

    negative_percentage = (
        negative_links / total_links
        if total_links > 0
        else 0
    )

    ax.text(
        0.03,
        0.97,
        window_name,
        transform=ax.transAxes,
        fontsize=35,
        verticalalignment='top',
        bbox=dict(
            boxstyle='round',
            facecolor='white',
            alpha=0.8
        )
    )

    ax.text(
        0.97,
        0.97,
        f'Negative links: {negative_percentage*100:.1f}%',
        transform=ax.transAxes,
        fontsize=35,
        verticalalignment='top',
        horizontalalignment='right',
        bbox=dict(
            boxstyle='round',
            facecolor='white',
            alpha=0.8
        )
    )

    ax.axis("off")

# =====================================================
# FINAL
# =====================================================

plt.tight_layout()
plt.show()
```

# PLOT UMBRAL P VALOR

```python
import numpy as np
import pickle
import matplotlib.pyplot as plt

# =====================================================
# ARCHIVOS
# =====================================================

files = {
    "w1": "gg_w1.pkl",
    "w2": "gg_w2.pkl",
    "w3": "gg_w3.pkl",
    "w4": "gg_w4.pkl"
}
files = {
    "w1": "gg_nopc_w1.pkl",
    "w2": "gg_nopc_w2.pkl",
    "w3": "gg_nopc_w3.pkl",
    "w4": "gg_nopc_w4.pkl"
}



files = {
    "w1": "results_w1_bm_pc.pkl",
    "w2": "results_w2_bm_pc.pkl",
    "w3": "results_w3_bm_pc.pkl"
}

files = {
    "w1": "mos_w1_par_nf.pkl",
    "w2": "mos_w2_par_nf.pkl",
    "w3": "mos_w3_par_nf.pkl",
    "w4": "mos_w4_par_nf.pkl"
}

# =====================================================
# RANGO DE P-VALORES
# =====================================================

p_thresholds = np.linspace(0.001, 0.5, 500)

# =====================================================
# FUNCIÓN
# =====================================================

def porcentaje_negativos(matriz_coef, matriz_pval, matriz_sig, p_th):

    n = matriz_coef.shape[0]
    
    total = 0
    negativos = 0

    for i in range(n):
        for j in range(n):

            if i == j:
                continue

            if (matriz_pval[i, j] < p_th):
                total += 1

                if matriz_coef[i, j] < 0:
                    negativos += 1

    if total == 0:
        return 0

    return negativos / total


# =====================================================
# CÁLCULO
# =====================================================

resultados = {}

for key, file in files.items():

    with open(file, "rb") as f:
        res = pickle.load(f)

    matriz_coef = res["matriz_coeficientes"]
    matriz_pval = res["matriz_pvalores"]
    matriz_sig= res["matriz_significancia"]

    porcentajes = []

    for p_th in p_thresholds:
        perc = porcentaje_negativos(matriz_coef, matriz_pval, matriz_sig, p_th)
        porcentajes.append(perc)

    resultados[key] = porcentajes


# =====================================================
# PLOT
# =====================================================

plt.figure(figsize=(7,4))

for key in resultados:
    plt.plot(p_thresholds, resultados[key], label=key)
plt.vlines(0.05,0,0.5, 'grey', "--", label="umbral p valor=0.05")
plt.xlabel("Umbral p-value")
plt.ylabel("Proporción de links negativos")
plt.ylim(0, 0.7)
plt.title("Con porcentajes")
plt.legend()

plt.show()
```
