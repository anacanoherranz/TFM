```python

def construir_enlaces_dirigidos(mat_sig, especies, species_to_community):
    enlaces = set()
    
    for i, sp_i in enumerate(especies):
        if sp_i not in species_to_community:
            continue
        ci = species_to_community[sp_i]
        
        for j, sp_j in enumerate(especies):
            if sp_j not in species_to_community:
                continue
            if i == j:
                continue
            
            cj = species_to_community[sp_j]
            val = mat_sig[i][j]
            
            if val == 0:
                continue
            
            # enlace dirigido con signo
            enlaces.add((ci, cj, int(val)))
    
    return enlaces

redes_dir = []


for data in datos:
    especies = data["especies"]
    mat_sig = data["matriz_significancia"]
    
    enlaces = construir_enlaces_dirigidos(mat_sig, especies, species_to_community)
    redes_dir.append(enlaces)
```

```python
import pickle
import pandas as pd
import numpy as np

files = ["results_w1_wio.pkl", "results_w2_wio.pkl", "results_w3_wio.pkl"]
datos = []
for f in files:
    with open(f, "rb") as file:
        datos.append(pickle.load(file))

groups_df = pd.read_csv("basa_groups.csv")
species_to_community = dict(zip(groups_df["Taxa"], groups_df["Community"]))

print("=== NIVEL ESPECIE ===")
for i, data in enumerate(datos, 1):
    especies = data["especies"]
    mat = np.array(data["matriz_significancia"])

    nodos = set()
    links_pos = 0
    links_neg = 0

    for r in range(len(especies)):
        for c in range(len(especies)):
            if r == c:
                continue
            val = mat[r][c]
            if val != 0:
                nodos.add(especies[r])
                nodos.add(especies[c])
                if val == 1:
                    links_pos += 1
                elif val == -1:
                    links_neg += 1

    print(f"Red {i}: nodos={len(nodos)}, links(+)={links_pos}, links(-)={links_neg}, total={links_pos + links_neg}, neg_ratio={links_neg/(links_pos + links_neg):.2f}")

print("\n=== NIVEL COMUNIDAD ===")
for i, data in enumerate(datos, 1):
    especies = data["especies"]
    mat = np.array(data["matriz_significancia"])

    # Cada enlace es una tripleta única (comunidad_origen, comunidad_destino, signo)
    # Máximo 4 enlaces por par de comunidades: (+1 y -1 en cada dirección)
    enlaces = set()
    nodos_com = set()

    for r in range(len(especies)):
        sp_r = especies[r]
        if sp_r not in species_to_community:
            continue
        ci = species_to_community[sp_r]

        for c in range(len(especies)):
            if r == c:
                continue
            sp_c = especies[c]
            if sp_c not in species_to_community:
                continue
            cj = species_to_community[sp_c]
            val = mat[r][c]
            if val == 0:
                continue

            enlaces.add((ci, cj, int(val)))
            nodos_com.add(ci)
            nodos_com.add(cj)

    links_pos = sum(1 for e in enlaces if e[2] == 1)
    links_neg = sum(1 for e in enlaces if e[2] == -1)
    print(f"Red {i}: nodos={len(nodos_com)}, links(+)={links_pos}, links(-)={links_neg}, total={len(enlaces)}, neg_ratio={links_neg/len(enlaces):.2f}")
```
